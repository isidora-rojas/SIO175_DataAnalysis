```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
M = 2
N = 10000

D = np.random.rand(M, N)
Davg = np.mean(D, axis=0)

plt.hist(Davg, density=True)
```

```python
M = 5
N = 10000

D = np.random.rand(M, N)
Davg = np.mean(D, axis=0)

plt.hist(Davg, density=True)
```

```python
M = 10
N = 10000

D = np.random.rand(M, N)
Davg = np.mean(D, axis=0)

plt.hist(Davg, density=True)
```

```python
mu = np.mean(Davg)
sigma = np.std(Davg)

x = np.linspace(0,1,1000)
y = norm.pdf(x, mu, sigma)

plt.hist(Davg,density=True)
plt.plot(x,y)
```

```python
for j in np.arange(2,20):
    D = np.random.rand(j, 10000)
    Davg = np.mean(D, axis=0)

    x = np.linspace(0,1,1000)
    y = norm.pdf(x, np.mean(Davg), np.std(Davg))

    plt.figure()
    plt.hist(Davg, density=True) 
    plt.plot(x,y)
    plt.title(f'M={j}')
```


## Square wave example

```matlab
N = 64;
dt = 1;
t = (0:N-1)'*dt;
x = [0;ones(N/2-1,1);0;-ones(N/2-1,1)];
figure (1)
clf
plot(t,x,'LineWidth',2)
title('Square wave')
ylim([-1.1 1.1])
grid on
```

### Square wave, cosine with 1 cycle/record length

```matlab
figure (2)
clf
plot(t,x,'LineWidth',2)
title('Square wave, cosine w/ 1 cycle/record length')
ylim([-1.1 1.1])
grid on
hold on
f = 1/(N*dt);
plot(t,cos(2*pi*f*t))
```

### Square wave, sine with 1 cycle/record length

```matlab
figure (3)
clf
plot(t,x,'LineWidth',2)
title('Square wave, sine w/ 1 cycle/record length')
ylim([-1.1 1.1])
grid on
hold on
f = 1/(N*dt);
plot(t,sin(2*pi*t/N))
```

### Square wave, sine with 2 cycles/record length

```matlab
figure (4)
clf
plot(t,x,'LineWidth',2)
title('Square wave, sine w/ 2 cycles/record length')
ylim([-1.1 1.1])
grid on
hold on
f = 2/(N*dt);
plot(t,sin(2*pi*f*t))
```

## Square wave, sine waves up to Nyquist

```matlab
N = 64;
y = zeros(N,1);
dt = 1;
f = zeros(N/2+1,1);
B = zeros(N/2+1,1);
A = zeros(N/2+1,1);
figure (5)
clf
for j = 1:N/2+1
    f(j) = (j-1)/(N*dt);
    if(j==1)
        A(j) = sum(x);
    else
        A(j) = 2*sum(x.*cos(2*pi*f(j)*t));
        B(j) = 2*sum(x.*sin(2*pi*f(j)*t));
    end
    y = y+B(j)/N*sin(2*pi*f(j)*t);
    plot(t,x,t,y,'--','LineWidth',2)
    ylim([-1.5 1.5])
    xlabel('time')
    ylabel('amplitude')
    grid on
    title(['j = ',num2str(j)])
    pause (.4)
end
```

### Square wave, complex exponential

```matlab
clf;
N = 64;
dt = 1;
X = zeros(N,1);
y = zeros(N,1);
i = sqrt(-1);
for j = 1:N
    f(j) = (j-1)/(N*dt);
    X(j) = sum(x.*exp(i*2*pi*f(j)*t));
    y = y+X(j)/N*exp(-i*2*pi*f(j)*t);
    plot(t,x,t,real(y),'--','LineWidth',2)
    ylim([-1.5 1.5])
    grid on
    title(['j = ',num2str(j-1)])
    pause (.4)
end
```
